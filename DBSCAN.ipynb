```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
from google.colab import files
import io

uploaded = files.upload()

for fn in uploaded.keys():
  print('User uploaded file "{name}" with length {length} bytes'.format(
      name=fn, length=len(uploaded[fn])))
  dataset = pd.read_csv(io.StringIO(uploaded[fn].decode('utf-8')))
```

```python
dataset
```

```python
X = dataset.iloc[:, 3:5].values
```

```python
X
```

```python
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors

# K-distance plot (elbow-equivalent for choosing eps)
neighbors = NearestNeighbors(n_neighbors=5)
neighbors_fit = neighbors.fit(X)
distances, indices = neighbors_fit.kneighbors(X)
distances = np.sort(distances[:, 4])
plt.plot(distances)
plt.title('K-distance Graph')
plt.xlabel('Points')
plt.ylabel('Epsilon')
plt.show()

dbscan = DBSCAN(eps=8, min_samples=5)
y_dbscan = dbscan.fit_predict(X)

y_dbscan

supervised = dataset
supervised['Cluster_group'] = y_dbscan
supervised.to_csv("cluster_dbscan.csv", index=False)
```

```python

import seaborn as sns
facet = sns.lmplot(data=supervised, x=supervised.columns[3], y=supervised.columns[4], hue=supervised.columns[5],
                   fit_reg=False, legend=True, legend_out=True)
```
